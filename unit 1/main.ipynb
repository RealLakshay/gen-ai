## Experiment 1: Text Generation


This experiment tests whether an architecture can perform Auto-regressive Generation.

```python
import torch
from transformers import pipeline

models = ["bert-base-uncased", "roberta-base", "facebook/bart-base"]
prompt = "The future of Artificial Intelligence is"

for model_name in models:
    print(f"\n--- Testing Generation with: {model_name} ---")
    try:
        # We use the 'text-generation' task
        generator = pipeline("text-generation", model=model_name)
        output = generator(prompt, max_length=20, num_return_sequences=1)
        print(output[0]['generated_text'])
    except Exception as e:
        print(f"FAILURE: {model_name} cannot perform standard generation easily.")
        print(f"Reason: {str(e)[:100]}...")
```

## Experiment 2: Masked Language Modeling (Fill-Mask)

This validates the Bi-directional Understanding capabilities of the models.

```python
# BERT and BART use [MASK], but RoBERTa uses <mask>
test_sentences = {
    "bert-base-uncased": "The goal of Generative AI is to [MASK] new content.",
    "roberta-base": "The goal of Generative AI is to <mask> new content.",
    "facebook/bart-base": "The goal of Generative AI is to [MASK] new content."
}

for model_name, sentence in test_sentences.items():
    print(f"\n--- Testing Fill-Mask with: {model_name} ---")
    try:
        filler = pipeline("fill-mask", model=model_name)
        results = filler(sentence)
        for res in results[:2]: # Top 2 predictions
            print(f"Predicted: '{res['token_str']}' (Score: {res['score']:.4f})")
    except Exception as e:
        print(f"Error with {model_name}: {e}")
```

## Experiment 3: Question Answering

The goal of this experiment was to evaluate the capability of Base Models (BERT, RoBERTa, and BART) to identify and extract an answer from a given context without task-specific fine-tuning.

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"

for model_name in models:
    print(f"\n--- Testing QA with: {model_name} ---")
    try:
        qa_pipeline = pipeline("question-answering", model=model_name)
        result = qa_pipeline(question=question, context=context)
        print(f"Answer: {result['answer']}")
    except Exception as e:
        print(f"Error with {model_name}: {e}")
```
