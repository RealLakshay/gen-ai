```python
from transformers import pipeline

def run_improved_detector():
    # 1. Initialize the classifier
    # We use top_k=None instead of return_all_scores=True for better compatibility
    detector = pipeline("text-classification", 
                        model="roberta-base-openai-detector", 
                        top_k=None)

    print("--- AI Bullshit Detector: Professional Version ---")
    print("Threshold set to 85% for high-confidence flagging.")
    
    while True:
        user_input = input("\n[Input Text] (or type 'exit'): ")
        
        if user_input.lower() == 'exit':
            break

        # 2. Get results
        # The pipeline returns a list of results (one per input)
        # Each result is a list of dictionaries [{'label': '...', 'score': ...}, ...]
        results = detector(user_input)[0] 
        
        # 3. Correctly extract scores using a dictionary comprehension
        scores = {res['label']: res['score'] for res in results}
        
        # Check for different possible label naming conventions
        fake_score = (scores.get('Fake') or scores.get('LABEL_0') or 0) * 100
        real_score = (scores.get('Real') or scores.get('LABEL_1') or 0) * 100

        print("-" * 40)
        
        if fake_score > 85:
            print(f"VERDICT: DETECTED BULLSHIT")
            print(f"Confidence: {fake_score:.2f}%")
            print("Action: High probability of misinformation or clickbait.")
        elif real_score > 85:
            print(f"VERDICT: LIKELY LEGITIMATE")
            print(f"Confidence: {real_score:.2f}%")
            print("Action: Text matches reliable reporting patterns.")
        else:
            print(f"VERDICT: UNCERTAIN / BORDERLINE")
            print(f"Real: {real_score:.2f}% | Fake: {fake_score:.2f}%")
            print("Action: AI is conflicted. Please perform a manual fact-check.")
            
        print("-" * 40)

if __name__ == "__main__":
    run_improved_detector()
```
